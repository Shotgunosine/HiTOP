```python
from pathlib import Path
import pandas as pd
```

# Before running
unzip `../data/archived_results.zip` to `../data/`

```python
bare_data = Path('../data')
bare_cfa_dir = bare_data / 'cfa'
bare_icc_dir = bare_data / 'icc'
bare_cv_dir = bare_data / 'convdiv'
og_data = Path('../data/archived_results/')
og_cfa_dir = og_data / 'cfa'
og_icc_dir = og_data / 'icc'
og_cv_dir = og_data / 'convdiv'
```

```python
bare_orig_cfa_res = pd.read_csv(bare_cfa_dir / 'orig_cfa_res.csv')
og_orig_cfa_res = pd.read_csv(og_cfa_dir / 'orig_cfa_res.csv')
assert bare_orig_cfa_res.equals(og_orig_cfa_res)
```

```python
b_stepwise_res = pd.read_pickle(bare_cfa_dir / 'stepwise.pkl')
o_stepwise_res = pd.read_pickle(og_cfa_dir / 'stepwise.pkl')
assert b_stepwise_res.equals(o_stepwise_res)
```

```python
b_exhaustive = pd.read_pickle(bare_cfa_dir / 'exhaustive.pkl')
o_exhaustive = pd.read_pickle(og_cfa_dir / 'exhaustive.pkl')
assert b_exhaustive.equals(o_exhaustive)
```

```python
b_other_inv = pd.read_pickle(bare_cfa_dir / 'other_scales_exhaustive.pkl')
o_other_inv = pd.read_pickle(og_cfa_dir / 'other_scales_exhaustive.pkl')
assert b_other_inv.equals(o_other_inv)
```

```python
b_cores = pd.read_pickle(bare_cfa_dir / 'consolidated_cores.pkl')
o_cores = pd.read_pickle(og_cfa_dir / 'consolidated_cores.pkl')
assert b_cores.equals(o_cores)
```

```python
b_core_icc = pd.read_pickle(bare_icc_dir / 'cores.pkl')
o_core_icc = pd.read_pickle(og_icc_dir / 'cores.pkl')
assert b_core_icc.equals(o_core_icc)
```

```python
file_names = [
    'result_ICC_origscales_gp.csv',
    'result_ICC_origscales_en.csv',
    'result_ICC_origscales_combined.csv',
    'result_ICC_origscales_forplot_mostcombined_indecivinessenriched.csv',
    'result_ICC_invariantonly_gp.csv',
    'result_ICC_invariantonly_en.csv',
    'result_ICC_invariantonly_combined.csv'
]

for fn in file_names:
    b_dat = pd.read_csv(bare_icc_dir / fn)
    o_dat = pd.read_csv(og_icc_dir / fn)
    assert b_dat.equals(o_dat)
```
